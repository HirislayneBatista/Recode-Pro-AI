# Treinamento, Avaliação e Métricas

• MSE: calcula o erro quadrático médio entre as previsões e os valores reais.

• RMSE (Root Mean Squared Error - Raiz do Erro Quadrático Médio): mede a magnitude média dos erros do modelo. Em outras palavras, ele te diz, em média, o quão distante as previsões do seu modelo estão dos valores reais. Dará uma ideia da magnitude dos erros. Útil para entender o quão "distantes" as previsões estão da realidade.

• R² (Coeficiente de Determinação): mede o quão bem o modelo explica a variação nos dados. Ele te diz a proporção da variância na variável dependente (preços, no seu caso), que é explicada pelas variáveis independentes (as features que você usou para fazer a previsão). Dará uma ideia de quão bem o modelo se ajusta aos dados. Útil para entender o poder explicativo do modelo.

##Exemplo 1 - Regressão: Previsão do Preço de Imóveis

###Treinamento

Escolha do Algoritmo: regressão Linear (simples e interpretável para este exemplo).

Divisão dos Dados: 70% para treinamento, 30% para teste (aleatoriamente).

Treinamento do Modelo: utilizado o conjunto de treinamento para ajustar os coeficientes da equação da reta, minimizando o erro quadrático médio (MSE).

##Avaliação

Previsões no Conjunto de Teste: utilizado o modelo treinado para prever os preços dos imóveis no conjunto de teste.

Comparação com Valores Reais: comparado as previsões com os preços reais no conjunto de teste.

##Métricas de Avaliação

MSE: cálculo do erro quadrático médio entre as previsões e os valores reais.

RMSE: cálculo da raiz quadrada do MSE para obter o erro na mesma unidade dos preços (R$).

R²: cálculo do coeficiente de determinação para avaliar o quão bem o modelo explica a variação nos preços.

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score


# Criar os dados
dados = {
    'Tamanho': [80, 120, 60, 100, 150],
    'NumeroQuartos': [2, 3, 1, 2, 4],
    'Idade_em_anos': [10, 5, 20, 15, 2],
    'Preco': [300000, 500000, 200000, 400000, 700000]
}
data = pd.DataFrame(dados)

# Separar as características (X) e o alvo (y)
X = data[['Tamanho', 'NumeroQuartos', 'Idade_em_anos']]
y = data['Preco']


# Dividir os dados em conjuntos de treinamento e teste
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)


# Criar e treinar o modelo
model = LinearRegression()
model.fit(X_train, y_train)


# Fazer previsões no conjunto de teste
y_pred = model.predict(X_test)


# Avaliar o modelo
mse = mean_squared_error(y_test, y_pred)
rmse = mse ** 0.5
r2 = r2_score(y_test, y_pred)
print("RMSE:", rmse)
print("R²:", r2)


# Exibindo as previsões em comparação com os valores reais
result_df = pd.DataFrame({'Valor Real (R$)': y_test, 'Previsão (R$)': y_pred})
print(result_df)
```

##Exemplo 2 - Classificação Binária: Previsão de Quitação de Empréstimo


###Treinamento
Escolha do Algoritmo: regressão Logística.

Pré-processamento: converte a coluna "Histórico de Crédito" em valores numéricos (ex: Bom = 1, Ruim = 0).

Divisão dos Dados: 70% para treinamento, 30% para teste (aleatoriamente).

Treinamento do Modelo: utilizado o conjunto de treinamento para ajustar os parâmetros do modelo, maximizando a capacidade de distinguir entre as duas classes (quitou ou não quitou).

###Avaliação
Previsões no Conjunto de Teste: utilizado o modelo treinado para prever se os clientes no conjunto de teste quitarão ou não o empréstimo.

Matriz de Confusão: construido uma matriz de confusão para visualizar os verdadeiros positivos, verdadeiros negativos, falsos positivos e falsos negativos.

###Métricas de Avaliação
Acurácia: cálculo da proporção de previsões corretas em relação ao total de previsões.

Precisão: cálculo da proporção de verdadeiros positivos em relação ao total de positivos previstos.

Recall: cálculo da proporção de verdadeiros positivos em relação ao total de positivos reais.

F1-Score: cálculo da média harmônica entre precisão e recall.

AUC: cálculo da área sob a curva ROC para avaliar a capacidade do modelo de distinguir entre as classes.

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report, precision_score, recall_score, f1_score, roc_auc_score, confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns

# Gerar dados
np.random.seed(42)
n_samples = 100

renda_mensal = np.random.randint(2000, 10001, size=n_samples)
historico_credito = np.random.choice(['Bom', 'Ruim'], size=n_samples, p=[0.6, 0.4])
quitou_emprestimo = np.random.choice(['Sim', 'Não'], size=n_samples, p=[0.7, 0.3])


dados = pd.DataFrame({
    'Renda Mensal (R$)': renda_mensal,
    'Histórico de Crédito': historico_credito,
    'Quitou o Empréstimo?': quitou_emprestimo
})
display(dados)
```

##Exemplo 3 - Classificação Multiclasse: Classificação de Frutas

###Treinamento
Escolha do Algoritmo: árvore de Decisão (simples e interpretável para este exemplo).

Pré-processamento: convertido as colunas categóricas ("Cor", "Forma", "Tamanho") em valores numéricos (ex: usando One-Hot Encoding ou Label Encoding).

Divisão dos Dados: 70% para treinamento, 30% para teste (aleatoriamente).

Treinamento do Modelo: utilizado o conjunto de treinamento para construir a árvore de decisão, maximizando a capacidade de distinguir entre as diferentes classes de frutas.

###Avaliação
Previsões no Conjunto de Teste: utilizado o modelo treinado para prever os tipos de frutas no conjunto de teste.

Matriz de Confusão: construido uma matriz de confusão para visualizar os acertos e erros do modelo para cada classe de fruta.

###Métricas de Avaliação
Acurácia: cálculo da proporção de previsões corretas em relação ao total de previsões.

Precisão, Recall e F1-Score: foi calculados para cada classe de fruta individualmente. Em seguida, podem ser agregados (macro-média, micro-média ou média ponderada).

```python
import pandas as pd
import numpy as np

# Define as possíveis características das frutas (dados não condiz com o real - é somente para realizar o treinamento do modelo)
cores = ['Vermelho', 'Amarelo', 'Laranja', 'Verde', 'Roxo']
formas = ['Redondo', 'Oval']
tamanhos = ['Pequeno', 'Médio', 'Grande']
tipos_fruta = ['Maçã', 'Banana', 'Laranja', 'Abacate', 'Uva']

# Cria uma lista para armazenar os dados
dados_frutas = []

# Gera 100 amostras aleatórias
for _ in range(200):
  cor = np.random.choice(cores)
  forma = np.random.choice(formas)
  tamanho = np.random.choice(tamanhos)
  tipo_fruta = np.random.choice(tipos_fruta)
  dados_frutas.append([cor, forma, tamanho, tipo_fruta])

# Cria um DataFrame pandas a partir dos dados gerados
df_frutas = pd.DataFrame(dados_frutas, columns=['Cor', 'Forma', 'Tamanho', 'Tipo de Fruta'])

# Exibe o DataFrame
display(df_frutas)

# Salva o DataFrame em um arquivo CSV
# df_frutas.to_csv('frutas_dataset.csv', index=False)
```
