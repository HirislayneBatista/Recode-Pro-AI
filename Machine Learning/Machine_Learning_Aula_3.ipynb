#Dataset Emprestimo Bancario

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report

# Carregar os dados (assumindo que estão em um arquivo CSV)
dados = pd.read_csv("/content/Dataset_de_Emprestimo_Bancario.csv")
display(dados)
```

```python
# Verificação de todos os valores únicos nas colunas
print(dados['Historico_Credito'].unique())
```

```python
# Conversão das características categóricas em numéricas
dados['Historico_Credito'] = dados['Historico_Credito'].map({'Ruim': 0, 'Bom': 1, 'Regular': 0.5})
```

```python
# Separação das Variáveis de predição e prevista
X = dados[['Idade', 'Renda_Mensal', 'Historico_Credito', 'Dividas']]
y = dados['Aprovado']

# Separação dos conjuntos de treinamento e teste
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Criação e Treinamento do modelo
modelo = LogisticRegression()
modelo.fit(X_train, y_train)

# Previsão pelo conjunto de teste
y_pred = modelo.predict(X_test)

# Resultado das previsões
result_df = pd.DataFrame({'Valor Real (R$)': y_test, 'Previsão (R$)': y_pred})
print(result_df)

# Validação
print("Acurácia:", accuracy_score(y_test, y_pred))
print("Relatório de Classificação:\n", classification_report(y_test, y_pred))
```
